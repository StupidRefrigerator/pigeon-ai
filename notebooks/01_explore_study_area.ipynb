## Fixed geographic decisions

The saved preliminary area reproduces at density 50, despite the former saved count/title indicating 500. Positional cluster IDs are unstable. Production now uses explicit approved preliminary SA1 membership, recalculates the named growth-suburb exclusions, and reports any difference from the approved final membership. The active urban-core threshold is 500; it does not silently redefine the approved boundary. The registry is a versioned geographic decision input, not a feature cache.

```python
from pigeon_ai.paths import RAW, REFERENCE

def read_ids(path):
    return set(path.read_text().splitlines())

# The pipeline no longer saves the preliminary area, so rebuild it from the registry.
sa1 = gpd.read_file(RAW / "abs_sa1_2021/SA1_2021_AUST_GDA2020.shp")
preliminary_ids = read_ids(REFERENCE / "approved_preliminary_sa1_ids.txt")
preliminary = sa1[sa1.SA1_CODE21.isin(preliminary_ids)].to_crs(CRS)
study = gpd.read_file(INTERIM / "boundaries/sydney_study_sa1_final.geojson")
print(f"Preliminary SA1s: {len(preliminary):,}")
print(f"Final study SA1s: {len(study):,}")
approved = read_ids(REFERENCE / "approved_final_sa1_ids.txt")
print("Matches approved final membership:", set(study.SA1_CODE21) == approved)
print("Final study population:", int(study.population.sum()))
```

```python
fig, ax = plt.subplots(figsize=(8, 8))
preliminary.plot(ax=ax, color="lightgrey", linewidth=0)
study.plot(ax=ax, color="steelblue", linewidth=0)
ax.set(title="Approved study area; excluded growth areas in grey")
ax.set_axis_off()
plt.show()
```
