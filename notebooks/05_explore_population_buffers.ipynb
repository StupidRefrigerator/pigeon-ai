# PIGEON-AI: contextual population and final checks

Population buffers use all intersecting original SA1s, including those outside the prediction area. Denominators remain complete original SA1 areas. Buffers overlap, so their populations cannot be summed as distinct regional residents.

The final table separates `hex_id`, the placement proxy `toilet_count`, 20 predictors and geometry. This notebook reads existing outputs and does not train a model.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "src").exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root / "src"))

import geopandas as gpd
import matplotlib.pyplot as plt
from pigeon_ai.paths import BASELINE, INTERIM, PROCESSED, REFERENCE
```

```python
from pigeon_ai.validation import FEATURES, compare_baseline, validate_features

features = gpd.read_file(PROCESSED / "sydney_hex_features.gpkg")
validate_features(features)
if (BASELINE / "sydney_hex_features.csv").exists():
    compare_baseline(features)
else:
    print("Skipped baseline comparison: archive/ not found.")
print(f"Predictors: {len(FEATURES)}")
display(features[["population_est", "population_250m", "population_500m"]].describe())
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(features["population_250m"], features["population_500m"], s=2, alpha=0.25)
ax.set_xlabel("Estimated residents within 250 m")
ax.set_ylabel("Estimated residents within 500 m")
ax.set_title("Nested population buffers")
plt.show()
```
